```python
from dataset import train_dataloader,test_dataloader
from model import TinyGpt
from config import VOCAB_PATH,MAX_SEQ_LEN,LR,EPOCHS,LOG_DIR,CHECKPOINTS_DIR
import torch

from utils import train
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
```

```python
loaded_vocab = torch.load(VOCAB_PATH,weights_only=True)

vocab = loaded_vocab['vocab']
merges = loaded_vocab['merges']
stoi = loaded_vocab['stoi']
itos = loaded_vocab['itos']
```
